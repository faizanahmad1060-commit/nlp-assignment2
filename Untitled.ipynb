```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

cats = ["rec.sport.hockey", "sci.space", "comp.graphics", "talk.politics.guns"]
data = fetch_20newsgroups(subset="train", categories=cats,
                          remove=("headers", "footers", "quotes"))
df = pd.DataFrame({"text": data.data})
df = df[df["text"].str.len() > 100]
df.to_csv("raw.csv", index=False)
print(df.shape)
print(df["text"].iloc[0][:300])
```

```python
from gensim.test.utils import datapath
import pandas as pd

p = datapath("lee_background.cor")
lines = [l.strip() for l in open(p, encoding="latin-1") if l.strip()]
df = pd.DataFrame({"text": lines})
df.to_csv("raw.csv", index=False)
print(df.shape)
print(df["text"].iloc[0][:300])
```
